# RANA AI — film in background mode (free Kaggle GPU, open source)

Renders the whole ad **without a browser**, so Kaggle's 40-minute idle timeout can't stop it.

**One-time settings (right-hand panel → Session options):** Accelerator **GPU T4 x2** · Internet **On**.

**Every new ad:**
1. Edit the brief in cell 1 (theme, tone, language, shots, lines).
2. Click **Save Version** (top right) → **Save & Run All (Commit)** → **Save**.
3. Close the tab if you like. It runs up to ~10.5 h in Kaggle's cloud.
4. When the version shows *Complete*: open it → **Output** tab → download **RANA_final.mp4** (and `RANA_final.srt`).

If something fails, `run.log`, `wangp.log` and `deepy_chat.json` in the Output tab show why.
Open source only: WanGP (Deepy Prime) · Qwen3.8 VL 9B · Wan 2.2 / LTX-2 · OmniVoice · FFmpeg.
